## 1. OpenAI-Compatible Chat Completion
Works seamlessly as a drop-in replacement by directing `base_url` to `https://api.batchin.tech/v1`.

````python
from openai import OpenAI
import os

# Set your API key
api_key = os.getenv("BATCHIN_API_KEY", "your-batchin-api-key")

client = OpenAI(
    base_url="https://api.batchin.tech/v1",
    api_key=api_key
)

# Invoke DeepSeek-V4 Pro
response = client.chat.completions.create(
    model="deepseek-v4-pro",
    messages=[{"role": "user", "content": "Explain how Hedged Dual-Dispatch eliminates P99 tail latency."}]
)

print(response.choices[0].message.content)
````

## 2. Zero-Token JSON Auto-Healer
Autonomous agents often fail when models emit truncated JSON. Watch the Auto-Healer repair broken structures client-side in microseconds without re-prompting.

````python
from batchin import JsonAutoHealer

# Simulating a truncated JSON emitted by an LLM hitting token limit
broken_llm_output = '''```json
{
  "status": "success",
  "decision": "approved",
  "actions": ["verify_merkle", "settle_x402",
'''

# Repairing without any additional LLM prompt calls
repaired_json = JsonAutoHealer.repair(broken_llm_output)
print("Repaired Deterministic JSON:")
print(repaired_json)
````
